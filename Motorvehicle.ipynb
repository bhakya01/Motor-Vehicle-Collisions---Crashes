Feature Engineeing

```python
import numpy as np
import pandas as pd
# First, parse the hour from CRASH TIME
crash_hour = pd.to_numeric(
    df['CRASH TIME'].str[:2],
    errors='coerce'
)

# Define the conditions and labels based on the extracted hour
conditions = [
    crash_hour.between(5, 11),
    crash_hour.between(12, 16),
    crash_hour.between(17, 20),
    (crash_hour.between(21, 23) | crash_hour.between(0, 4))
]

labels = ['Morning', 'Afternoon', 'Evening', 'Night']

# Apply the selection to create the TIME PERIOD column
df['TIME PERIOD'] = np.select(
    conditions, labels, default='Unknown'
)

# Display the counts of each time period
print(df['TIME PERIOD'].value_counts())
```

```python
conditions = [
    df['NUMBER OF PERSONS KILLED'] > 0,
    df['NUMBER OF PERSONS INJURED'] > 0
]

labels = ['Fatal', 'Injury']

df['CRASH SEVERITY'] = np.select(
    conditions,
    labels,
    default='No Reported Injury or Death'
)

print(df['CRASH SEVERITY'].value_counts())
```

```python
import pandas as pd

# Convert CRASH DATE to datetime format (DD-MM-YYYY)
df['CRASH DATE'] = pd.to_datetime(df['CRASH DATE'], format='%d-%m-%Y', errors='coerce')

# Extract year and month
df['CRASH YEAR'] = df['CRASH DATE'].dt.year
df['CRASH MONTH'] = df['CRASH DATE'].dt.month
df['CRASH DAY'] = df['CRASH DATE'].dt.day
# Display the extracted columns alongside the original date
display(df[['CRASH DATE', 'CRASH YEAR', 'CRASH MONTH', 'CRASH DAY']].head())
```
