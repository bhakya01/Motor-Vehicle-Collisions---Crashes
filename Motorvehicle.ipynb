**Data Cleaning**

**Feature Engineeing (EDA)**

```python
df.rename(columns={
    'Unnamed: 21': 'Vechicle name 2',
    'VEHICLE TYPE CODE 4': 'Vechicle name 1'
}, inplace=True)

print(df.columns)
```

**Afternoon Commute Vulnerability:** The Afternoon period (12:00 PM to 4:59 PM) has the highest concentration of crashes with 5,809 incidents, closely followed by the Morning period (5:00 AM to 11:59 AM) with 5,718 incidents.  Together, these daytime/commuting periods account for over 55% of all crashes.

**Lower Risk at Twilight:** The Evening window (5:00 PM to 8:59 PM) shows the lowest risk with 4,296 crashes, indicating that while traffic density is high, the shorter duration of this transition block lowers absolute crash counts.

**Nighttime Disproportion:** Despite significantly lower road occupancy overnight, the Night period (9:00 PM to 4:59 AM) still records a substantial 4,824 crashes, highlighting factors like reduced visibility, fatigue, or speed during off-peak hours.

```python

# First, parse the hour from CRASH TIME
crash_hour = pd.to_numeric(
    df['CRASH TIME'].str[:2],
    errors='coerce'
)

# Define the conditions and labels based on the extracted hour
conditions = [
    crash_hour.between(5, 11),
    crash_hour.between(12, 16),
    crash_hour.between(17, 20),
    (crash_hour.between(21, 23) | crash_hour.between(0, 4))
]

labels = ['Morning', 'Afternoon', 'Evening', 'Night']

# Apply the selection to create the TIME PERIOD column
df['TIME PERIOD'] = np.select(
    conditions, labels, default='Unknown'
)

# Display the counts of each time period
print(df['TIME PERIOD'].value_counts())
```

```python
conditions = [
    df['NUMBER OF PERSONS KILLED'] > 0,
    df['NUMBER OF PERSONS INJURED'] > 0
]

labels = ['Fatal', 'Injury']

df['CRASH SEVERITY'] = np.select(
    conditions,
    labels,
    default='No Reported Injury or Death'
)

print(df['CRASH SEVERITY'].value_counts())
```

```python

# Convert CRASH DATE to datetime format (DD-MM-YYYY)
df['CRASH DATE'] = pd.to_datetime(df['CRASH DATE'], format='%d-%m-%Y', errors='coerce')

# Extract year and month
df['CRASH YEAR'] = df['CRASH DATE'].dt.year
df['CRASH MONTH'] = df['CRASH DATE'].dt.month
df['CRASH DAY'] = df['CRASH DATE'].dt.day

# Display the extracted columns alongside the original date
display(df[['CRASH DATE', 'CRASH YEAR', 'CRASH MONTH', 'CRASH DAY']].head(10))
```

**DATA VISUALIZATIONS**


**Extreme Concentration in 2024:** The vast majority of vehicle crashes in this dataset occurred in 2024 (comprising over 99% of the records).

**Data Completion Dropoff:** The years 2025 and 2026 show exceptionally low crash counts (dropping down to single digits). Rather than a miraculous reduction in accidents, this points to a data limitation where the dataset only covers a tiny, incomplete portion of early 2025/2026.

**Planning Implications:** For accurate predictive modeling or trend analysis, you should filter your analyses specifically to 2024 to avoid seasonal skewing caused by the incomplete tracking in subsequent years

```python
df['CRASH YEAR'].value_counts().sort_index().plot(kind='bar', figsize=(8,5))

plt.title('Number of Crashes by Year')
plt.xlabel('Crash Year')
plt.ylabel('Crash Count')
plt.show()
```

**Most Vulnerable Season:** Accident counts peak noticeably during the late spring and early summer months **(May and June)** as well as the autumn/pre-holiday season **(October and November)**. This correlation often relates to increased outdoor recreational activities, road trips, and changing weather/daylight hours.

**Lower Accident Counts:** **February and January** typically show a lower absolute number of recorded crashes. While winter weather poses driving challenges, reduced outdoor activity, shorter daylight hours, and proactive road salt treatments often keep total recreational road travel lower.

**Planning Recommendations:** Traffic monitoring and prevention programs should strategically allocate higher patrol presence during peak spring and autumn months to counter these seasonal spikes.

**pivot table:**

**Most Dangerous Collision Dynamics:** The highest volume of motorist injuries occurs in collisions involving Station Wagon/Sport Utility Vehicles colliding with other Station Wagon/Sport Utility Vehicles **(1,211 injuries) or with** Sedans

**The Common Factor:** **Because Sedans and Station Wagons/SUVs** make up the  their mutual collisions dominate the overall safety metrics, producing the largest clusters of motorist injuries in the entire dataset.


```python
# Remove extra spaces from column names
df.columns = df.columns.str.strip()

# Convert injury column to numeric
df['NUMBER OF MOTORIST INJURED'] = pd.to_numeric(
    df['NUMBER OF MOTORIST INJURED'],
    errors='coerce'
)

# Get top 10 vehicle types
top1 = df['Vechicle name 1'].value_counts().head(10).index
top2 = df['Vechicle name 2'].value_counts().head(10).index

# Filter data
filtered = df[
    df['Vechicle name 1'].isin(top1) &
    df['Vechicle name 2'].isin(top2)
]

# Create pivot table
injury_pivot = pd.pivot_table(
    filtered,
    index='Vechicle name 1',
    columns='Vechicle name 2',
    values='NUMBER OF MOTORIST INJURED',
    aggfunc='sum',
    fill_value=0
)

# Heatmap
plt.figure(figsize=(12,8))

sns.heatmap(
    injury_pivot,
    annot=True,
    fmt='.0f',
    cmap='YlOrRd'
)

plt.title('Motorist Injuries: Vehicle 1 vs Vehicle 2')
plt.xlabel('Vehicle Name 2')
plt.ylabel('Vehicle Name 1')
plt.show()
```

```python
df['NUMBER OF MOTORIST KILLED'] = pd.to_numeric(
    df['NUMBER OF MOTORIST KILLED'],
    errors='coerce'
)

killed_group = df.groupby(
    ['Vechicle name 1', 'Vechicle name 2']
)['NUMBER OF MOTORIST KILLED'].sum().reset_index()

killed_group = killed_group.sort_values(
    by='NUMBER OF MOTORIST KILLED',
    ascending=False
)

print(killed_group.head(10))
```

**BAR CHART**

**1. Highest Fatalities:** Sedan + Station Wagon/Sport Utility Vehicle recorded the highest number, with 5 motorist deaths.

**2. Other Vehicle Combinations:** Tractor Truck Diesel + Motorcycle recorded 1 death.

**3. Motorcycle-related Fatalities:** Motorscooter + Bus and Motorcycle + Tractor Truck Diesel each recorded 1 death.

**4. Low Fatality Counts:** Most of the displayed vehicle combinations recorded only 1 death, while some recorded zero.

**Draw back in Sedan**
Possible Risk Factors:Differences in vehicle size, weight, visibility, collision speed and impact direction may influence crash severity.

**Recommendation: Improve:** driver awareness, maintain safe following distances, follow speed limits and promote the use of seatbelts.

```python
top10 = killed_group.head(10).copy()

top10['Vehicle Combination'] = (
    top10['Vechicle name 1'].astype(str) + ' + ' +
    top10['Vechicle name 2'].astype(str)
)

plt.figure(figsize=(12,6))

sns.barplot(
    data=top10,
    x='NUMBER OF MOTORIST KILLED',
    y='Vehicle Combination',
    color='salmon'
)

plt.title('Top 10 Vehicle Combinations by Motorist Deaths')
plt.xlabel('Total Motorists Killed')
plt.ylabel('Vehicle Combination')
plt.show()
```

**Line chart**

**1. Highest Death Count**: January 2024 recorded the highest daily motorist death count, with 5 deaths on a single date.

**2. Low Daily Fatalities**: Most recorded dates show 0 or 1 motorist death.

**3. 2025–2026 Trend**: The graph shows mostly zero deaths on the recorded dates during 2025 and early 2026.

```python
df['CRASH DATE'] = pd.to_datetime(
    df['CRASH DATE'], errors='coerce', dayfirst=True
)

daily_deaths = df.groupby('CRASH DATE')[
    'NUMBER OF MOTORIST KILLED'
].sum()

plt.figure(figsize=(14,6))

plt.plot(
    daily_deaths.index,
    daily_deaths.values,
    marker='o'
)

plt.title('Daily Motorist Deaths Trend')
plt.xlabel('Crash Date')
plt.ylabel('Number of Motorists Killed')
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```
