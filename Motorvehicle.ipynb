**Feature Engineeing (EDA)**

```python
import numpy as np
import pandas as pd
# First, parse the hour from CRASH TIME
crash_hour = pd.to_numeric(
    df['CRASH TIME'].str[:2],
    errors='coerce'
)

# Define the conditions and labels based on the extracted hour
conditions = [
    crash_hour.between(5, 11),
    crash_hour.between(12, 16),
    crash_hour.between(17, 20),
    (crash_hour.between(21, 23) | crash_hour.between(0, 4))
]

labels = ['Morning', 'Afternoon', 'Evening', 'Night']

# Apply the selection to create the TIME PERIOD column
df['TIME PERIOD'] = np.select(
    conditions, labels, default='Unknown'
)

# Display the counts of each time period
print(df['TIME PERIOD'].value_counts())
```

```python
conditions = [
    df['NUMBER OF PERSONS KILLED'] > 0,
    df['NUMBER OF PERSONS INJURED'] > 0
]

labels = ['Fatal', 'Injury']

df['CRASH SEVERITY'] = np.select(
    conditions,
    labels,
    default='No Reported Injury or Death'
)

print(df['CRASH SEVERITY'].value_counts())
```

**DATA VISUALIZATIONS**


```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# ==========================================
# UNIVARIATE ANALYSIS
# NYC MOTOR VEHICLE CRASH DATASET
# ==========================================

# 1. Crash Severity Distribution
if 'CRASH SEVERITY' in df.columns:
    severity_counts = df['CRASH SEVERITY'].value_counts()

    print("Crash Severity Distribution:")
    print(severity_counts)

    severity_counts.plot(kind='bar', figsize=(9, 5))
    plt.title('Univariate Analysis: Crash Severity')
    plt.xlabel('Crash Severity')
    plt.ylabel('Number of Crashes')
    plt.xticks(rotation=25, ha='right')
    plt.tight_layout()
    plt.show()


# 2. Crashes by Time Period
if 'TIME PERIOD' in df.columns:
    period_order = ['Morning', 'Afternoon', 'Evening', 'Night']

    period_counts = (
        df['TIME PERIOD']
        .value_counts()
        .reindex(period_order, fill_value=0)
    )

    print("\nCrashes by Time Period:")
    print(period_counts)

    period_counts.plot(kind='bar', figsize=(8, 5))
    plt.title('Univariate Analysis: Crash Time Period')
    plt.xlabel('Time Period')
    plt.ylabel('Number of Crashes')
    plt.xticks(rotation=0)
    plt.tight_layout()
    plt.show()


# 3. Crashes by Borough
if 'BOROUGH' in df.columns:
    borough = df['BOROUGH'].astype('string').str.strip()
    borough = borough.replace(
        ['', 'unknown', 'Unknown', 'UNKNOWN', 'nan'],
        pd.NA
    ).dropna()

    print("\n Crashes by Borough:")
    print(borough.value_counts().head(10))

    borough.value_counts().head(10).plot(
        kind='bar', figsize=(9, 5)
    )
    plt.title('Univariate Analysis: Crashes by Borough')
    plt.xlabel('Borough')
    plt.ylabel('Number of Crashes')
    plt.xticks(rotation=30, ha='right')
    plt.tight_layout()
    plt.show()


# 4. Top 10 Contributing Factors
factor_col = 'CONTRIBUTING FACTOR VEHICLE 1'

if factor_col in df.columns:
    factors = df[factor_col].astype('string').str.strip()
    factors = factors.replace(
        ['', 'unknown', 'Unknown', 'UNKNOWN', 'nan'],
        pd.NA
    ).dropna()

    top_factors = factors.value_counts().head(10)

    print("\nTop 10 Contributing Factors:")
    print(top_factors)

    top_factors.plot(kind='barh', figsize=(10, 6))
    plt.title('Univariate Analysis: Top Contributing Factors')
    plt.xlabel('Number of Crashes')
    plt.ylabel('Contributing Factor')
    plt.gca().invert_yaxis()
    plt.tight_layout()
    plt.show()


# 5. Crashes by Month
if 'CRASH MONTH' in df.columns:
    month = pd.to_numeric(
        df['CRASH MONTH'], errors='coerce'
    ).dropna()

    print("\nCrashes by Month:")
    print(month.value_counts().sort_index())

    month.value_counts().sort_index().plot(
        kind='bar', figsize=(9, 5)
    )
    plt.title('Univariate Analysis: Crashes by Month')
    plt.xlabel('Month')
    plt.ylabel('Number of Crashes')
    plt.xticks(
        ticks=range(12),
        labels=range(1, 13),
        rotation=0
    )
    plt.tight_layout()
    plt.show()


# 6. Distribution of Crash Time (Hour)
if 'CRASH TIME' in df.columns:
    hour = pd.to_numeric(
        df['CRASH TIME'].astype('string').str[:2],
        errors='coerce'
    )
    hour = hour[hour.between(0, 23)]

    print("\nCrashes by Hour:")
    print(hour.value_counts().sort_index())

    hour.value_counts().sort_index().reindex(
        range(24), fill_value=0
    ).plot(kind='line', marker='o', figsize=(10, 5))

    plt.title('Univariate Analysis: Crashes by Hour')
    plt.xlabel('Hour of Day (24-hour format)')
    plt.ylabel('Number of Crashes')
    plt.xticks(range(24))
    plt.grid(True)
    plt.tight_layout()
    plt.show()


# 7. Distribution of Persons Injured
injury_col = 'NUMBER OF PERSONS INJURED'

if injury_col in df.columns:
    injured = pd.to_numeric(
        df[injury_col], errors='coerce'
    ).dropna()

    print("\nPersons Injured - Summary:")
    print(injured.describe())

    plt.figure(figsize=(8, 5))
    plt.hist(injured, bins=30, edgecolor='black')
    plt.title('Univariate Analysis: Distribution of Persons Injured')
    plt.xlabel('Number of Persons Injured per Crash')
    plt.ylabel('Frequency')
    plt.tight_layout()
    plt.show()

print("\nUNIVARIATE ANALYSIS COMPLETED!")
```

```python

# LOCATION (coordinates) vs CRASH SEVERITY

lat = pd.to_numeric(df['LATITUDE'], errors='coerce')
lon = pd.to_numeric(df['LONGITUDE'], errors='coerce')

valid_coordinates = (
    lat.between(40.4, 41.0) &
    lon.between(-74.3, -73.6)
)

location_df = df.loc[
    valid_coordinates,
    ['CRASH SEVERITY']
].copy()

# Approximate location cells
location_df['LAT_CELL'] = lat[valid_coordinates].round(2)
location_df['LON_CELL'] = lon[valid_coordinates].round(2)

location_severity = pd.crosstab(
    [location_df['LAT_CELL'], location_df['LON_CELL']],
    location_df['CRASH SEVERITY']
)

print("Top 10 Location Cells by Crash Count")
location_severity['TOTAL'] = location_severity.sum(axis=1)

display(
    location_severity
    .sort_values('TOTAL', ascending=False)
    .head(10)
)

# Plot the 10 location cells with the most crashes
top_locations = (
    location_severity
    .sort_values('TOTAL', ascending=False)
    .head(10)
    .drop(columns='TOTAL')
)

top_locations.plot(
    kind='bar',
    stacked=True,
    figsize=(12, 6)
)

plt.title('Crash Severity by Approximate Geographic Location')
plt.xlabel('Latitude and Longitude Cell')
plt.ylabel('Number of Crashes')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

### Combined Geographical & Behavioral Multivariate Analysis
This analysis combines localized coordinates, ZIP codes, and vehicle contributing factors to pinpoint where and why high-severity crashes are concentrated.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Clean geographic and factor columns
df['ZIP CODE'] = df['ZIP CODE'].replace(0, pd.NA)
lat = pd.to_numeric(df['LATITUDE'], errors='coerce')
lon = pd.to_numeric(df['LONGITUDE'], errors='coerce')
valid_coordinates = lat.between(40.4, 41.0) & lon.between(-74.3, -73.6)

# Prepare subset for multidimensional grouping
geo_behavior_df = df.loc[valid_coordinates, [
    'ZIP CODE', 'LATITUDE', 'LONGITUDE',
    'CONTRIBUTING FACTOR VEHICLE 1',
    'CONTRIBUTING FACTOR VEHICLE 2',
    'CRASH SEVERITY'
]].copy()

geo_behavior_df['LAT_CELL'] = lat[valid_coordinates].round(2)
geo_behavior_df['LON_CELL'] = lon[valid_coordinates].round(2)

# Replace uninformative labels with NA for behavioral variables
for col in ['CONTRIBUTING FACTOR VEHICLE 1', 'CONTRIBUTING FACTOR VEHICLE 2']:
    geo_behavior_df[col] = geo_behavior_df[col].replace(
        ['unknown', 'Unknown', 'UNKNOWN', 'Unspecified', ''], pd.NA
    )

# Filter out completely uninformative rows for the behavioral interaction
cleaned_behavior = geo_behavior_df.dropna(subset=['CONTRIBUTING FACTOR VEHICLE 1', 'CONTRIBUTING FACTOR VEHICLE 2'])

# Get top 5 ZIP codes and top 5 Vehicle 1 factors
top_zips = cleaned_behavior['ZIP CODE'].value_counts().head(5).index
top_factors1 = cleaned_behavior['CONTRIBUTING FACTOR VEHICLE 1'].value_counts().head(5).index

filtered_subset = cleaned_behavior[
    cleaned_behavior['ZIP CODE'].isin(top_zips) &
    cleaned_behavior['CONTRIBUTING FACTOR VEHICLE 1'].isin(top_factors1)
]

# 2. Pivot Table: ZIP Code vs. Vehicle 1 Factor vs. Vehicle 2 Factor (by Crash Count)
zip_factor_matrix = pd.crosstab(
    [filtered_subset['ZIP CODE'], filtered_subset['CONTRIBUTING FACTOR VEHICLE 1']],
    filtered_subset['CONTRIBUTING FACTOR VEHICLE 2']
)

print("=== MULTIVARIATE MATRIX: ZIP CODE + FACTOR 1 VS FACTOR 2 ===")
display(zip_factor_matrix)

# 3. Create Heatmap for Geographic Cells vs top contributing behaviors
top_locations = cleaned_behavior.groupby(['LAT_CELL', 'LON_CELL']).size().sort_values(ascending=False).head(5).index
geo_factor_subset = cleaned_behavior[
    cleaned_behavior.set_index(['LAT_CELL', 'LON_CELL']).index.isin(top_locations) &
    cleaned_behavior['CONTRIBUTING FACTOR VEHICLE 1'].isin(top_factors1)
]

loc_factor_pivot = pd.crosstab(
    [geo_factor_subset['LAT_CELL'], geo_factor_subset['LON_CELL']],
    geo_factor_subset['CONTRIBUTING FACTOR VEHICLE 1']
)

plt.figure(figsize=(12, 6))
sns.heatmap(loc_factor_pivot, annot=True, cmap='YlOrRd', fmt='d', cbar_kws={'label': 'Crash Count'})
plt.title('Multivariate Analysis: Top Geographic Hotspots vs. Top Primary Contributing Factors')
plt.xlabel('Primary Contributing Factor')
plt.ylabel('Geographic Coordinate Cells (Lat, Lon)')
plt.xticks(rotation=30, ha='right')
plt.tight_layout()
plt.show()
```
